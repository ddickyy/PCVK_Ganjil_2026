# STEP 2
Baca image dari URL

````python
# Membuat list menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
      ]
# baca dan tampilkan gambar
# loop pada tiap url, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)   # baca image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) # resize image hhalf size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) # convert image to RGB
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
````

# STEP 3

````python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("Resolusi image: tinggi x lebar = ", tinggi, "x", lebar)
cv2_imshow(image_2)
````

# STEP 4
Mengakses piksel dengan garis diagonal menyilang

````python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat garis horizontal di tengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2,image_3))
cv2_imshow(final_frame)
````

**Studi Kasus 1**: Panitia sebuah kegiatan perlu membubuhkan label identitas pada ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel, sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga proporsinya tetap sama.


````python
from google.colab.patches import cv2_imshow
def tambah_label(img, teks):
    h, w = img.shape[:2]

    bar_h = int(h * 0.12)

    bar = np.zeros((bar_h, w, 3), dtype=np.uint8)

    font_scale = h / 700
    thickness = max(1, int(h / 250))

    (text_w, text_h), _ = cv2.getTextSize(teks,cv2.FONT_HERSHEY_SIMPLEX,font_scale,thickness)

    x = (w - text_w) // 2
    y = (bar_h + text_h) // 2

    cv2.putText(bar, teks, (x, y), cv2.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), thickness)
    hasil = np.vstack((img, bar))

    return hasil

img1 = cv2.imread("/content/WhatsApp Image 2026-08-31 at 11.17.29 AM.jpeg")
img2 = cv2.imread("/content/WhatsApp Image 2026-08-31 at 5.53.49 PM.jpeg")
img1_rgb = img1.copy()
img1_rgb = cv2.cvtColor(img1_rgb, cv2.COLOR_BGR2RGB)
img2_rgb = img2.copy()
img2_rgb = cv2.cvtColor(img2_rgb, cv2.COLOR_BGR2RGB)

hasil1 = tambah_label(img1_rgb, "Foto KTM")
hasil2 = tambah_label(img2_rgb, "Dokumentasi PBL")

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(hasil1)
plt.title("Foto Pertama")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(hasil2)
plt.title("Foto Kedua")
plt.axis('off')

plt.show()
````

**Studi Kasus 2**: Marketplace seperti Tokopedia dan Shopee menampilkan foto produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja. Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna. Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas, tambahkan logo nama toko berukuran proporsional pada gambar.

````python
def buat_tumbnail(img, nama_toko):
    h, w = img.shape[:2]
    sisi = max(h, w)

    kanvas = np.zeros((sisi, sisi, 3), dtype=np.uint8)
    kanvas[:] = (240, 240, 240)

    if w > h:
        new_w = sisi
        new_h = int(h * sisi / w)
    else:
        new_h = sisi
        new_w = int(w * sisi / h)

    foto = cv2.resize(img, (new_w, new_h))

    x = (sisi - new_w) // 2
    y = (sisi - new_h) // 2

    kanvas[y:y+new_h, x:x+new_w] = foto
    font_scale = sisi / 700
    thickness = max(1, int(sisi / 250))

    (text_w, text_h), _ = cv2.getTextSize(nama_toko,cv2.FONT_HERSHEY_SIMPLEX,font_scale,thickness)

    text_x = (sisi - text_w) // 2
    text_y = sisi - text_h - int(sisi * 0.03)

    cv2.putText(kanvas,nama_toko,(text_x, text_y),cv2.FONT_HERSHEY_SIMPLEX,font_scale,(0, 0, 0),thickness)
    return kanvas

img = io.imread("https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg")

hasil = buat_tumbnail(img, "Warung California")
cv2_imshow(hasil)
````
